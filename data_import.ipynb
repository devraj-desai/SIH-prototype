```python
import copernicusmarine

copernicusmarine.subset(
  dataset_id="cmems_mod_glo_phy_my_0.083deg_P1M-m",
  variables=["bottomT", "mlotst", "siconc", "sithick", "so", "thetao", "uo", "usi", "vo", "vsi", "zos"],
  minimum_longitude=63.52076479165809,
  maximum_longitude=92.85108934522951,
  minimum_latitude=-2.179680173519668,
  maximum_latitude=23.534576969337472,
  start_datetime="2026-05-01T00:00:00",
  end_datetime="2026-06-01T00:00:00",
  minimum_depth=0.49402499198913574,
  maximum_depth=1062.43994140625,
)
```

```python
import xarray as xr

# Open your NetCDF file
ds = xr.open_dataset("data.nc")

# Inspect available variables
print(ds)

# Select one variable (e.g., temperature)
df = ds["thetao"].to_dataframe().reset_index()

# Save to CSV
df.to_csv("output.csv", index=False)
```

```python
for t in ds.time:
    df = ds.sel(time=t, depth=0, method="nearest")[["thetao", "so", "uo", "vo"]].to_dataframe().reset_index()
    df.to_csv(f"marine_{str(t.values)[:10]}.csv", index=False)
```
